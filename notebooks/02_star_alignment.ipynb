# Step 02: STAR alignment


Aligns each sample to GRCh38.p14 with `--quantMode GeneCounts`. Outputs BAM and `ReadsPerGene.out.tab` to `Data/AlignBAM/<sample>/`.


This is the most time-consuming step (~10 min per sample). It runs in the background; monitor progress with `!tail -f Data/logs/02_star_align.log`.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd()))

import yaml
from utils.helpers import launch_background

cfg = yaml.safe_load(open('config.yaml'))
launch_background(
    [sys.executable, 'scripts/star_align.py',
     '--samplesheet', 'Data/samplesheet.csv',
     '--genome-dir', cfg['star']['genome_dir'],
     '--out-dir', 'Data/AlignBAM',
     '--threads', str(cfg['star']['threads']),
     '--quant-mode', cfg['star']['quant_mode'],
     '--star-bin', cfg['tools']['star_bin']],
    'Data/logs/02_star_align.log'
)
```
